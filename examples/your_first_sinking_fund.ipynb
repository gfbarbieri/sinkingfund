# Your First Sinking Fund

Welcome to the Sinking Fund library! This tutorial will guide you through creating your first sinking fund from scratch. By the end, you'll have:

- Created a single bill
- Created an envelope for that bill
- Generated a contribution schedule
- Produced your first report

This is a minimal example designed for maximum clarity. We'll use the simplest possible setup to help you understand the core concepts.


## Step 1: Import Required Modules

First, we need to import the necessary components. The `SinkingFund` class is the main interface that orchestrates everything.


````python
from datetime import date
from sinkingfund import SinkingFund
````

## Step 2: Create Your Sinking Fund

A sinking fund needs a planning period (start and end dates) and an initial balance. Let's create a fund for the year 2025 with $1,000 to start.


````python
# Create a sinking fund for 2025 with $1,000 initial balance.
fund = SinkingFund(
    start_date=date(2025, 1, 1),
    end_date=date(2025, 12, 31),
    balance=1000.00
)

print(f"Planning period: {fund.start_date} to {fund.end_date}")
print(f"Initial balance: ${fund.balance}")
````

## Step 3: Define Your First Bill

A bill represents a financial obligation you need to save for. Let's create a simple one-time bill: property tax due on November 1st, 2025 for $3,600.

We'll define it as a dictionary with the required fields:
- `bill_id`: A unique identifier
- `service`: A descriptive name
- `amount_due`: The amount needed
- `recurring`: Whether it repeats (False for one-time bills)
- `due_date`: When the payment is due


````python
# Define a one-time property tax bill.
property_tax_bill = {
    "bill_id": "prop_tax_2025",
    "service": "Property Tax 2025",
    "amount_due": 3600.00,
    "recurring": False,
    "due_date": date(2025, 11, 1)
}

# Add the bill (automatically creates envelope).
# add_bills() accepts a single dict, list of dicts, or file path.
fund.add_bills(property_tax_bill)

print("Bill created successfully!")
print(f"Bill ID: {property_tax_bill['bill_id']}")
print(f"Service: {property_tax_bill['service']}")
print(f"Amount due: ${property_tax_bill['amount_due']}")
print(f"Due date: {property_tax_bill['due_date']}")
````

## Step 4: Generate Your First Report

Now comes the magic! The `quick_report()` method does everything for you in one call:

1. **Creates envelopes** for all bills in your planning period
2. **Allocates your initial balance** to those envelopes
3. **Sets up contribution schedules** (we'll use bi-weekly contributions)
4. **Generates a daily account report** showing balances over time

Let's generate a report with bi-weekly contributions (every 14 days):


````python
# Generate a quick report with bi-weekly contributions.
report = fund.quick_report(contribution_interval=14)

print(f"Report generated for {len(report)} days")
print(f"First date: {min(report.keys())}")
print(f"Last date: {max(report.keys())}")
````

## Step 5: Understanding Your Report

The report is a dictionary where each key is a date, and each value contains information about that day. Let's examine a few key dates to understand what the report tells us.


````python
# Look at the first day of the planning period.
first_date = date(2025, 1, 1)
first_day = report[first_date]

print(f"=== Report for {first_date} ===")
print(f"Account Balance: ${first_day['account_balance']['total']}")
print(f"Contributions: ${first_day['contributions']['total']}")
print(f"Payouts: ${first_day['payouts']['total']}")
print()
print("Breakdown by bill:")
for bill_id, amount in first_day['account_balance']['bills'].items():
    print(f"  {bill_id}: ${amount}")
````

Let's check a day when a contribution occurs (every 14 days starting from the planning start date):


````python
# Check a contribution day (January 15th, 14 days after start).
contribution_date = date(2025, 1, 15)
contribution_day = report[contribution_date]

print(f"=== Report for {contribution_date} ===")
print(f"Account Balance: ${contribution_day['account_balance']['total']}")
print(f"Contributions: ${contribution_day['contributions']['total']}")
print(f"Payouts: ${contribution_day['payouts']['total']}")
print()
if contribution_day['contributions']['total'] > 0:
    print("Contributions breakdown:")
    for bill_id, amount in contribution_day['contributions']['bills'].items():
        if amount > 0:
            print(f"  {bill_id}: ${amount}")
````

Now let's check the due date to see what happens when the bill is paid:


````python
# Check the due date (November 1st).
due_date = date(2025, 11, 1)
due_day = report[due_date]

print(f"=== Report for {due_date} (Bill Due Date) ===")
print(f"Account Balance: ${due_day['account_balance']['total']}")
print(f"Contributions: ${due_day['contributions']['total']}")
print(f"Payouts: ${due_day['payouts']['total']}")
print()
if due_day['payouts']['total'] > 0:
    print("Payouts breakdown:")
    for bill_id, amount in due_day['payouts']['bills'].items():
        if amount > 0:
            print(f"  {bill_id}: ${amount}")
````

## Step 6: Visualizing Your Progress

Let's create a simple summary showing how your sinking fund grows over time. We'll look at key metrics across the planning period:


````python
# Find all contribution days and calculate totals.
contribution_days = []
total_contributed = 0
total_paid_out = 0

for report_date, daily_data in report.items():
    if daily_data['contributions']['total'] > 0:
        contribution_days.append(report_date)
        total_contributed += daily_data['contributions']['total']
    if daily_data['payouts']['total'] > 0:
        total_paid_out += daily_data['payouts']['total']

print("=== Sinking Fund Summary ===")
print(f"Total contributions: ${total_contributed:.2f}")
print(f"Total payouts: ${total_paid_out:.2f}")
print(f"Number of contribution days: {len(contribution_days)}")
print(f"Average contribution per day: ${total_contributed / len(contribution_days):.2f}")
print()
print("First few contribution dates:")
for d in contribution_days[:5]:
    print(f"  {d}: ${report[d]['contributions']['total']:.2f}")
````

## Step 7: Managing Your Sinking Fund

The SinkingFund API provides methods to update and manage your bills and balance:

### Updating Bills

You can update bill properties, and the system will automatically sync the associated envelopes:

```python
# Update the bill amount.
fund.update_bill("prop_tax_2025", {"amount_due": 3800.00})

# Update multiple properties.
fund.update_bill("prop_tax_2025", {
    "amount_due": 3800.00,
    "service": "Property Tax 2025 (Updated)"
})
```

### Updating Balance

When your account balance changes, you can update it:

```python
# Update balance after receiving funds.
fund.update_balance(2000.00)

# Regenerate report with new balance.
new_report = fund.quick_report(contribution_interval=14)
```

### Generating Reports with Different Options

You can generate reports with different strategies and options:

```python
# Generate report with different allocation strategy.
new_report = fund.quick_report(
    allocation_strategy="proportional",
    contribution_interval=7,  # Weekly instead of bi-weekly
    active_only=True
)
```

### State Management

Keep your sinking fund state consistent:

```python
# Validate state consistency.
is_valid, issues = fund.validate_state()
if not is_valid:
    for issue in issues:
        print(f"Issue: {issue}")

# Sync envelopes with bills (removes orphaned envelopes).
fund.sync_envelopes_with_bills()
```
