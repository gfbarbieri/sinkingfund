# Understanding Bills vs. Envelopes

One of the most important concepts in the Sinking Fund library is understanding the difference between **Bills** and **Envelopes**. These are distinct but related concepts that work together to help you plan and save for future expenses.

This tutorial will explain:
- What bills represent
- What envelopes represent
- How they relate to each other
- When to use each concept
- Common mistakes and how to avoid them


## What is a Bill?

A **Bill** represents a financial obligation - something you need to pay for in the future. Bills define:

- **What** you're paying for (service name)
- **How much** you need (amount due)
- **When** it's due (due date or schedule)
- **Whether** it repeats (recurring or one-time)

Think of a bill as the "contract" or "obligation" itself. For example:
- "I owe $3,600 for property tax on November 1st"
- "I pay $750 for car insurance every 6 months starting April 24th"

Bills are **definitions** - they describe what you need to pay, but they don't track how you're saving for it.


## What is an Envelope?

An **Envelope** represents a savings container for a specific bill instance. Envelopes track:

- **How much** you've allocated (initial allocation)
- **How much** you're contributing over time (contribution schedule)
- **When** you can contribute (contribution window)
- **Current balance** at any point in time

Think of an envelope as a physical envelope where you put money to save for a specific bill. For example:
- An envelope for "Property Tax due November 1st" that currently has $1,200 saved
- An envelope for "Car Insurance due April 24th" that receives $125 every two weeks

Envelopes are **execution** - they track how you're actually saving money to meet your bills.


## The Relationship: Bills → Bill Instances → Envelopes

Here's how they connect:

1. **Bill** (definition): "Property tax, $3,600, due annually on November 1st"
2. **Bill Instance** (specific occurrence): "Property tax due November 1, 2025 for $3,600"
3. **Envelope** (savings container): A container saving money specifically for that November 1, 2025 payment

For recurring bills, one Bill definition can create multiple Bill Instances, and each instance gets its own Envelope.

Let's see this in action:


````python
from datetime import date
from sinkingfund import SinkingFund, Bill

# Create a sinking fund.
fund = SinkingFund(
    start_date=date(2025, 1, 1),
    end_date=date(2025, 12, 31),
    balance=2000.00
)
````

### Example 1: One-Time Bill

A one-time bill creates exactly one bill instance, which gets one envelope:


````python
# Create a one-time bill.
one_time_bill = {
    "bill_id": "prop_tax",
    "service": "Property Tax 2025",
    "amount_due": 3600.00,
    "recurring": False,
    "due_date": date(2025, 11, 1)
}

# add_bills() automatically creates envelopes for bill instances in the planning period.
fund.add_bills([one_time_bill])

# Get bill instances in our planning period.
instances = fund.get_bill_instances()
print(f"Number of bill instances: {len(instances)}")
print(f"\nBill Instance:")
print(f"  Bill ID: {instances[0].bill_id}")
print(f"  Service: {instances[0].service}")
print(f"  Amount: ${instances[0].amount_due}")
print(f"  Due Date: {instances[0].due_date}")

# Envelopes are automatically created by add_bills(), but we can also check them.
envelopes = fund.get_envelopes()

print(f"\nNumber of envelopes: {len(envelopes)}")
print(f"\nEnvelope (automatically created):")
print(f"  Bill ID: {envelopes[0].bill_instance.bill_id}")
print(f"  Target Amount: ${envelopes[0].bill_instance.amount_due}")
print(f"  Initial Allocation: ${envelopes[0].initial_allocation}")
````

### Example 2: Recurring Bill

A recurring bill can create multiple bill instances, each getting its own envelope:


````python
# Create a new fund for this example.
fund2 = SinkingFund(
    start_date=date(2025, 1, 1),
    end_date=date(2026, 12, 31),  # Two-year period to see multiple instances
    balance=2000.00
)

# Create a recurring bill (every 6 months).
recurring_bill = {
    "bill_id": "car_insurance",
    "service": "Car Insurance",
    "amount_due": 750.00,
    "recurring": True,
    "start_date": date(2025, 4, 24),
    "frequency": "monthly",
    "interval": 6  # Every 6 months
}

# add_bills() automatically creates envelopes for all bill instances.
fund2.add_bills([recurring_bill])

# Get all bill instances in our planning period.
instances = fund2.get_bill_instances()
print(f"Number of bill instances: {len(instances)}")
print("\nBill Instances:")
for i, instance in enumerate(instances, 1):
    print(f"  {i}. {instance.service} - ${instance.amount_due} due {instance.due_date}")

# Envelopes are automatically created by add_bills().
envelopes = fund2.get_envelopes()

print(f"\nNumber of envelopes (automatically created): {len(envelopes)}")
print("\nEnvelopes:")
for i, envelope in enumerate(envelopes, 1):
    print(f"  {i}. {envelope.bill_instance.service}")
    print(f"     Due: {envelope.bill_instance.due_date}")
    print(f"     Target: ${envelope.bill_instance.amount_due}")
````

## Side-by-Side Comparison

Let's create a comprehensive example showing bills and envelopes working together:


````python
# Create a fund with multiple bills.
fund3 = SinkingFund(
    start_date=date(2025, 1, 1),
    end_date=date(2025, 12, 31),
    balance=3000.00
)

# Add both one-time and recurring bills.
bills = [
    {
        "bill_id": "prop_tax",
        "service": "Property Tax",
        "amount_due": 3600.00,
        "recurring": False,
        "due_date": date(2025, 11, 1)
    },
    {
        "bill_id": "car_ins",
        "service": "Car Insurance",
        "amount_due": 750.00,
        "recurring": True,
        "start_date": date(2025, 4, 24),
        "frequency": "monthly",
        "interval": 6
    }
]

fund3.add_bills(bills)

# Show the relationship.
print("=== BILLS (Definitions) ===")
all_bills = fund3.get_bills()
for bill in all_bills:
    print(f"\nBill ID: {bill.bill_id}")
    print(f"  Service: {bill.service}")
    print(f"  Amount: ${bill.amount_due}")
    print(f"  Recurring: {bill.recurring}")

print("\n=== BILL INSTANCES (Specific Occurrences) ===")
instances = fund3.get_bill_instances()
for instance in instances:
    print(f"\n{instance.service}")
    print(f"  Due: {instance.due_date}")
    print(f"  Amount: ${instance.amount_due}")

print("\n=== ENVELOPES (Savings Containers) ===")
# Envelopes are automatically created by add_bills(), but we can access them directly.
envelopes = fund3.get_envelopes()
for envelope in envelopes:
    print(f"\n{envelope.bill_instance.service}")
    print(f"  Due: {envelope.bill_instance.due_date}")
    print(f"  Target: ${envelope.bill_instance.amount_due}")
    print(f"  Allocated: ${envelope.initial_allocation}")
````

## When to Use Bills vs. Envelopes

### You work with Bills when:
- **Defining your financial obligations** - "What do I need to pay?"
- **Loading data from files** - CSV, Excel, JSON
- **Setting up your sinking fund** - Initial configuration
- **Understanding your expenses** - What bills are coming up?

### You work with Envelopes when:
- **Allocating funds** - How much money goes to each bill?
- **Creating contribution schedules** - When and how much to contribute?
- **Tracking progress** - How much is saved for each bill?
- **Checking funding status** - Will I have enough by the due date?

### The Library Handles the Connection

The `add_bills()` method automatically:
1. Creates bills from your data
2. Generates bill instances for the planning period
3. Creates envelopes for those instances
4. Sets up contribution dates

Then `quick_report()` automatically:
1. Allocates funds to envelopes
2. Creates contribution schedules
3. Generates reports

You can also manually work with envelopes using `create_envelopes()` if needed for advanced customization, but in most cases, `add_bills()` handles everything automatically.


## Common Mistakes and How to Avoid Them

### Mistake 1: Confusing Bills with Envelopes

**Wrong thinking**: "I need to put money into my bill"
**Correct thinking**: "I need to put money into the envelope for my bill"

Bills are definitions; envelopes are containers. You save money in envelopes, not bills.

### Mistake 2: Creating Envelopes Manually When You Don't Need To

**Wrong approach**: Manually creating envelopes for every bill instance
**Correct approach**: Let `add_bills()` automatically create envelopes, or use `create_envelopes()` if needed

```python
# Don't do this manually unless you need custom logic:
# envelope = Envelope(...)  # Usually unnecessary

# Do this instead:
# Envelopes are automatically created by add_bills()
# Or you can manually create them: fund.create_envelopes(instances)
```

### Mistake 3: Expecting One Bill = One Envelope

**Wrong assumption**: Each bill always creates exactly one envelope
**Correct understanding**: Recurring bills create multiple instances, each with its own envelope

A bill that occurs 4 times in your planning period will create 4 envelopes.

### Mistake 4: Modifying Bills Instead of Envelopes

**Wrong approach**: Trying to change allocation by modifying the bill
**Correct approach**: Work with envelopes for allocation and scheduling

Bills define obligations; envelopes handle savings execution.


## Practical Example: The Complete Flow

Let's see the complete flow from bills to envelopes to report:


````python
# Step 1: Define bills (what you owe).
fund4 = SinkingFund(
    start_date=date(2025, 1, 1),
    end_date=date(2025, 12, 31),
    balance=2000.00
)

fund4.add_bills([
    {
        "bill_id": "prop_tax",
        "service": "Property Tax",
        "amount_due": 3600.00,
        "recurring": False,
        "due_date": date(2025, 11, 1)
    }
])

print("Step 1: Bills defined")
print(f"  Number of bills: {len(fund4.get_bills())}")

# Step 2: Get bill instances (specific occurrences).
instances = fund4.get_bill_instances()
print(f"\nStep 2: Bill instances created")
print(f"  Number of instances: {len(instances)}")

# Step 3: Envelopes are automatically created by add_bills() (savings containers).
envelopes = fund4.get_envelopes()
print(f"\nStep 3: Envelopes automatically created")
print(f"  Number of envelopes: {len(envelopes)}")

# Step 4: Allocate funds to envelopes.
# First, set an allocation strategy.

fund4.allocate(strategy='sorted', sort_key='cascade')
print(f"\nStep 4: Funds allocated")
for envelope in envelopes:
    print(f"  {envelope.bill_instance.service}: ${envelope.initial_allocation} allocated")

# Step 5: Generate report (shows envelope balances over time).
report = fund4.quick_report(contribution_interval=14)
print(f"\nStep 5: Report generated")
print(f"  Report covers {len(report)} days")
print(f"  First day balance: ${report[date(2025, 1, 1)]['account_balance']['total']}")
````

## Summary

**Bills** = Financial obligations (what you need to pay)
- Define once
- Describe the obligation
- Create bill instances automatically

**Envelopes** = Savings containers (how you're saving)
- Created for each bill instance
- Track allocations and contributions
- Show current balance over time

**The Flow**:
1. Define Bills → 2. `add_bills()` automatically creates Bill Instances and Envelopes → 3. Allocate Funds → 4. Generate Schedules → 5. Create Reports

Remember: Bills tell you **what** you owe; Envelopes show you **how** you're saving for it!

**Note**: `add_bills()` automatically creates envelopes for you. You can also manually create them using `create_envelopes()` if needed for advanced workflows.
