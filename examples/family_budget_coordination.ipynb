# Family Budget Coordination

Managing a family budget means juggling 20+ bills across different categories: utilities, subscriptions, insurance, taxes, and more. This notebook demonstrates how to coordinate a comprehensive family budget, balancing urgent expenses with long-term goals, and generating reports that the whole family can understand.

**Topics covered:**
- Managing 20+ bills across categories
- Balancing urgent vs. long-term goals
- Generating family-friendly reports


```python
from datetime import date
from sinkingfund import SinkingFund
```

## Setting Up Family Bills

Let's create a comprehensive family budget with bills across multiple categories: utilities, subscriptions, insurance, taxes, and savings goals.


```python
# Comprehensive family budget with multiple bill categories.
family_bills = [
    # Utilities (recurring monthly)
    {'bill_id': 'electric', 'service': 'Electric Bill', 'amount_due': 150.00, 'recurring': True, 'start_date': date(2025, 1, 15), 'frequency': 'monthly', 'interval': 1, 'occurrences': 12},
    {'bill_id': 'water', 'service': 'Water & Sewer', 'amount_due': 80.00, 'recurring': True, 'start_date': date(2025, 1, 10), 'frequency': 'monthly', 'interval': 1, 'occurrences': 12},
    {'bill_id': 'gas', 'service': 'Natural Gas', 'amount_due': 120.00, 'recurring': True, 'start_date': date(2025, 1, 20), 'frequency': 'monthly', 'interval': 1, 'occurrences': 12},
    {'bill_id': 'internet', 'service': 'Internet Service', 'amount_due': 75.00, 'recurring': True, 'start_date': date(2025, 1, 1), 'frequency': 'monthly', 'interval': 1, 'occurrences': 12},
    
    # Subscriptions (recurring monthly)
    {'bill_id': 'netflix', 'service': 'Netflix', 'amount_due': 15.99, 'recurring': True, 'start_date': date(2025, 1, 5), 'frequency': 'monthly', 'interval': 1, 'occurrences': 12},
    {'bill_id': 'spotify', 'service': 'Spotify Premium', 'amount_due': 9.99, 'recurring': True, 'start_date': date(2025, 1, 8), 'frequency': 'monthly', 'interval': 1, 'occurrences': 12},
    {'bill_id': 'gym', 'service': 'Gym Membership', 'amount_due': 50.00, 'recurring': True, 'start_date': date(2025, 1, 1), 'frequency': 'monthly', 'interval': 1, 'occurrences': 12},
    
    # Insurance (annual)
    {'bill_id': 'home_insurance', 'service': 'Homeowners Insurance', 'amount_due': 1800.00, 'recurring': False, 'due_date': date(2025, 4, 1)},
    {'bill_id': 'car_insurance', 'service': 'Auto Insurance', 'amount_due': 1200.00, 'recurring': False, 'due_date': date(2025, 8, 15)},
    {'bill_id': 'life_insurance', 'service': 'Life Insurance Annual', 'amount_due': 600.00, 'recurring': False, 'due_date': date(2025, 10, 1)},
    
    # Taxes
    {'bill_id': 'property_tax', 'service': 'Property Tax', 'amount_due': 3600.00, 'recurring': False, 'due_date': date(2025, 11, 1)},
    
    # Annual subscriptions
    {'bill_id': 'software_sub', 'service': 'Software Subscription', 'amount_due': 599.00, 'recurring': False, 'due_date': date(2025, 6, 1)},
    {'bill_id': 'amazon_prime', 'service': 'Amazon Prime Annual', 'amount_due': 139.00, 'recurring': False, 'due_date': date(2025, 2, 15)},
    
    # Savings goals (treated as bills)
    {'bill_id': 'vacation_fund', 'service': 'Vacation Fund Goal', 'amount_due': 3000.00, 'recurring': False, 'due_date': date(2025, 7, 1)},
    {'bill_id': 'emergency_fund', 'service': 'Emergency Fund Goal', 'amount_due': 5000.00, 'recurring': False, 'due_date': date(2025, 12, 31)},
]

# Create family sinking fund.
family_fund = SinkingFund(
    start_date=date(2025, 1, 1),
    end_date=date(2025, 12, 31),
    balance=1000.00  # Starting emergency buffer
)

family_fund.create_bills(family_bills)

print(f"Family budget loaded: {len(family_fund.bill_manager.bills)} bills")
print(f"\nBill categories:")
print(f"  Recurring monthly: {len([b for b in family_fund.bill_manager.bills if b.recurring])}")
print(f"  One-time/annual: {len([b for b in family_fund.bill_manager.bills if not b.recurring])}")
```

## Generating the Family Budget Plan

We'll generate a comprehensive plan that balances urgent expenses (due soon) with long-term goals (end of year).


```python
# Generate family budget plan with bi-weekly contributions.
family_report = family_fund.quick_report(contribution_interval=14)

# Calculate totals.
total_bills = sum(bill.amount_due * (bill.occurrences if bill.recurring else 1) 
                  for bill in family_fund.bill_manager.bills)
total_contributions = sum(float(data['contributions']['total']) for data in family_report.values())

print("=== Family Budget Overview ===\n")
print(f"Planning period: {family_fund.start_date} to {family_fund.end_date}")
print(f"Starting balance: ${family_fund.balance:,.2f}")
print(f"Total obligations: ${total_bills:,.2f}")
print(f"Total contributions planned: ${total_contributions:,.2f}")
print(f"Bi-weekly contribution average: ${total_contributions / 26:,.2f}")
```

## Balancing Urgent vs. Long-Term Goals

The system automatically prioritizes urgent expenses (earliest due dates) while still planning for long-term goals. Let's see how this works.


```python
# Analyze priority allocation.
print("=== Expense Priority Analysis ===\n")

# Get all instances to see the full timeline.
instances = family_fund.get_bills_in_range()

# Group by urgency (Q1, Q2, Q3, Q4).
quarters = {
    'Q1 (Urgent)': [i for i in instances if i.due_date <= date(2025, 3, 31)],
    'Q2': [i for i in instances if date(2025, 4, 1) <= i.due_date <= date(2025, 6, 30)],
    'Q3': [i for i in instances if date(2025, 7, 1) <= i.due_date <= date(2025, 9, 30)],
    'Q4 (Long-term)': [i for i in instances if i.due_date >= date(2025, 10, 1)],
}

for quarter, quarter_instances in quarters.items():
    total = sum(float(inst.amount_due) for inst in quarter_instances)
    count = len(quarter_instances)
    print(f"{quarter}: ${total:,.2f} across {count} bills")
```

## Generating Family-Friendly Reports

Let's create a simplified monthly summary that's easy for the whole family to understand.


```python
# Generate family-friendly monthly summary.
from collections import defaultdict

monthly_summary = defaultdict(lambda: {'contributions': 0, 'bills_due': [], 'account_balance': 0})

# Group by month.
for report_date, data in family_report.items():
    month_key = report_date.strftime('%B %Y')
    monthly_summary[month_key]['contributions'] += float(data['contributions']['total'])
    monthly_summary[month_key]['account_balance'] = float(data['account_balance']['total'])
    
    # Track bills due this month.
    if data['payouts']['total'] < 0:
        # Find which bills are due around this date.
        for instance in instances:
            if instance.due_date == report_date:
                monthly_summary[month_key]['bills_due'].append(instance)

print("=== Monthly Family Budget Summary ===\n")

for month in sorted(monthly_summary.keys(), key=lambda x: date(2025, ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'].index(x.split()[0]) + 1, 1)):
    summary = monthly_summary[month]
    if summary['contributions'] > 0 or summary['bills_due']:
        print(f"{month}:")
        print(f"  Contributions: ${summary['contributions']:,.2f}")
        if summary['bills_due']:
            print(f"  Bills due:")
            for bill in summary['bills_due']:
                print(f"    - {bill.service}: ${bill.amount_due:,.2f}")
        print(f"  Account balance end of month: ${summary['account_balance']:,.2f}")
        print()
```

```python
# Organize bills by category.
categories = {
    'Utilities': ['electric', 'water', 'gas', 'internet'],
    'Subscriptions': ['netflix', 'spotify', 'gym', 'software_sub', 'amazon_prime'],
    'Insurance': ['home_insurance', 'car_insurance', 'life_insurance'],
    'Taxes': ['property_tax'],
    'Savings Goals': ['vacation_fund', 'emergency_fund']
}

print("=== Spending by Category ===\n")

for category, bill_ids in categories.items():
    category_total = 0
    category_bills = []
    
    for bill in family_fund.bill_manager.bills:
        if bill.bill_id in bill_ids:
            if bill.recurring:
                category_total += bill.amount_due * bill.occurrences
            else:
                category_total += bill.amount_due
            category_bills.append(bill.service)
    
    print(f"{category}:")
    print(f"  Total: ${category_total:,.2f}")
    print(f"  Bills: {', '.join(category_bills)}")
    print()
```
