# BillManager: Loading and Managing Multiple Bills

The `BillManager` centralizes bill operations, handling loading from multiple sources and generating bill instances for planning periods. This notebook demonstrates how to effectively use the BillManager to manage your bill collections.

**Topics covered:**
- Loading bills from multiple sources (CSV, Excel, JSON, dictionaries)
- Generating instances for planning periods
- Managing bill collections and validation


```python
from datetime import date
from sinkingfund.managers import BillManager
```

## Loading Bills from Multiple Sources

BillManager can load bills from files (CSV, Excel, JSON) or from Python dictionaries. The `create_bills()` method handles format detection automatically.


```python
# Create a BillManager.
manager = BillManager()

# Load bills from dictionary data (you can also use CSV, Excel, or JSON files).
sample_bill_data = [
    {
        'bill_id': 'prop_tax',
        'service': 'Property Tax',
        'amount_due': 3600.00,
        'recurring': True,
        'start_date': date(2025, 11, 1),
        'frequency': 'annual',
        'interval': 1
    },
    {
        'bill_id': 'car_ins',
        'service': 'Car Insurance',
        'amount_due': 750.00,
        'recurring': True,
        'start_date': date(2025, 4, 24),
        'frequency': 'monthly',
        'interval': 6
    }
]

bills_from_dict = manager.create_bills(sample_bill_data)
manager.add_bills(bills_from_dict)

print(f"Loaded {manager.get_bill_count()} bills")
print(f"Bill IDs: {[bill.bill_id for bill in manager.bills]}")
```

### Loading from Dictionaries

You can also create bills from Python dictionaries:


```python
# Create bills from dictionary data.
bill_data = [
    {
        'bill_id': 'netflix',
        'service': 'Netflix Subscription',
        'amount_due': 15.99,
        'recurring': True,
        'start_date': date(2025, 1, 1),
        'frequency': 'monthly',
        'interval': 1
    },
    {
        'bill_id': 'annual_fee',
        'service': 'Annual Membership Fee',
        'amount_due': 120.00,
        'recurring': False,
        'due_date': date(2025, 6, 1)
    }
]

bills_from_dict = manager.create_bills(bill_data)
manager.add_bills(bills_from_dict)

print(f"Total bills managed: {manager.get_bill_count()}")
```

## Generating Instances for Planning Periods

The `active_instances_in_range()` method generates all bill instances within a specified date range. This includes all instances in the range plus one "next" instance per bill for forward-looking planning.


```python
# Generate bill instances for 2025.
instances = manager.active_instances_in_range(
    start_reference=date(2025, 1, 1),
    end_reference=date(2025, 12, 31)
)

print(f"Generated {len(instances)} bill instances for 2025")
print("\nFirst 10 instances:")
for i, instance in enumerate(instances[:10], 1):
    print(f"  {i}. {instance.service} - ${instance.amount_due} due {instance.due_date}")
```

## Managing Bill Collections

BillManager provides methods to add and remove bills, with automatic validation to prevent duplicates.


```python
# Check current bill count.
print(f"Current bills: {manager.get_bill_count()}")

# Remove a bill by ID.
manager.remove_bill('netflix')
print(f"After removal: {manager.get_bill_count()}")

# List remaining bill IDs.
print(f"Remaining bills: {[bill.bill_id for bill in manager.bills]}")
```

## Summary

**Key Points:**

1. **Multi-Source Loading**: Use `create_bills()` with file paths (CSV, Excel, JSON) or dictionary lists
2. **Bill Management**: Use `add_bills()` and `remove_bill()` to manage your collection
3. **Instance Generation**: `active_instances_in_range()` creates all instances for planning periods
4. **Validation**: BillManager automatically validates and prevents duplicate bill IDs
